### Measuring Computation Time (CPU vs. GPU)

```python
print('--- CPU Computation ---')

# Create two random tensors on CPU
cpu_tensor1 = torch.rand(10000, 10000)
cpu_tensor2 = torch.rand(10000, 10000)

start_time_cpu = time.time()
# Perform matrix multiplication on CPU
cpu_result = torch.matmul(cpu_tensor1, cpu_tensor2)
end_time_cpu = time.time()

print(f"CPU computation time: {end_time_cpu - start_time_cpu:.4f} seconds")
```

```python
if torch.cuda.is_available():
    print('\n--- GPU Computation ---')
    # Create two random tensors on GPU
    gpu_tensor1 = torch.rand(10000, 10000, device=device)
    gpu_tensor2 = torch.rand(10000, 10000, device=device)

    start_time_gpu = time.time()
    # Perform matrix multiplication on GPU
    gpu_result = torch.matmul(gpu_tensor1, gpu_tensor2)
    torch.cuda.synchronize() # Wait for GPU to finish
    end_time_gpu = time.time()

    print(f"GPU computation time: {end_time_gpu - start_time_gpu:.4f} seconds")
else:
    print('\nGPU is not available, skipping GPU computation.')
```
